```python
import numpy as np
import pandas as pd
print(np.__version__)
print(pd.__version__)
```

```python
m=np.array([10,20,60,89,90])
print("Marks",m)
print("Type",type(m))
```

```python
print("Number of dimensions:",m.ndim)
print("Shape:",m.shape)
print("number of elements",m.size)
```

```python
m=np.array([10,20,60,89,90])
print("original",m)
print("Add 5",m+5)
print("Multiply by 2:",m*2)
print("Divide by 2:",m/2)
print("Square:",m**2)
```

```python
print("Sum:",m.sum())
print("Mean",m.mean())
print("Maximum:",m.max())
print("Minimum:",m.min())
```

```python
print("First marks:",m[0])
print("Second mark:",m[1])
print("last mark:",m[-1])
```

```python
print("Index 1 to 3:",m[1:4])
print("First three",m[:3])
print("Every second element:",m[::2])
```

```python
s=np.array([
    [10,20,30,40],
    [33,43,54,65],
    [12,43,54,77]
])
print(s)
print("Shape:",s.shape)
```

```python
print("Mean of all values:",s.mean())
print("Column wise mean:",s.mean(axis=0))
print("Row wise mean",s.mean(axis=1))
```
